```python
import matplotlib.pyplot as plt
import numpy as np

baseline_speed = 0.7254
indexer_speed = [0.5274, 0.4280, 0.4273, 0.4230]
memory_speed = [0.6722, 0.6508, 0.6410, 0.6329]

cache_memory = [7.6794, 6.6784, 5.6775, 4.6766, 4.0760]
cache_labels = ["0.0", "0.25", "0.50", "0.75", "0.90"]

cr_labels = ["0.25", "0.50", "0.75", "0.90"]

x = np.arange(len(cr_labels))
width = 0.25

fig, axes = plt.subplots(1, 2, figsize=(8, 4), dpi=200)

# Speed subplot
ax = axes[0]
ax.bar(x - width / 2, indexer_speed, width, label="Indexer only", color="#98CDFA")
ax.bar(x + width / 2, memory_speed, width, label="IndexMem", color="#5F4CD5")
ax.axhline(baseline_speed, linestyle="--", color="gray", label="baseline")
ax.set_xticks(x)
ax.set_xticklabels(cr_labels)
ax.set_xlabel("Compression Ratio")
ax.set_ylabel("Average decoding time (s)")
ax.set_ylim(top=0.85)
ax.legend()

# Cache memory subplot
ax = axes[1]
bars = ax.bar(cache_labels, cache_memory, color="#ACCEFB")
for bar, mem in zip(bars, cache_memory):
    ax.text(bar.get_x() + bar.get_width() / 2, bar.get_height(), f"{mem:.2f}", 
            ha='center', va='bottom', fontsize=9)
ax.set_xlabel("Compression Ratio")
ax.set_ylabel("Cache memory (GB)")

plt.tight_layout()
plt.savefig("speed_and_memory.png", dpi=400)
plt.show()
```
